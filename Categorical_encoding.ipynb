CATEGORICAL ENCODING :
Categorical Encoding is used to convert categorical (text-based) data into numerical data so that machine learning algorithms can process it.

```python
# Method 1: Label Encoding
# Label Encoding is a categorical encoding technique that converts text-based categorical values into numerical values.
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Creating dataset
data = {
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha', 'Rohit', 'Anjali', 'Karan', 'Neha'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'City': ['Delhi', 'Mumbai', 'Indore', 'Delhi', 'Mumbai', 'Indore', 'Delhi', 'Mumbai'],
    'Education': ['Graduate', 'Postgraduate', 'Undergraduate', 'Graduate',
                  'Postgraduate', 'Undergraduate', 'Graduate', 'Postgraduate'],
    'Purchased': ['Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)



print(df)
print("______________________________________________________________________________________________________________________")

#Identify Categorical Columns
cat_cols = df.select_dtypes(include='object').columns

print(cat_cols)
print("______________________________________________________________________________________________________________________")


# Creating LabelEncoder object
le = LabelEncoder()

# Applying Label Encoding
df['Purchased'] = le.fit_transform(df['Purchased'])

print(df)
print("______________________________________________________________________________________________________________________")

# Inverse Transform
df['Purchased'] = le.inverse_transform(df['Purchased'])

print(df)
print("______________________________________________________________________________________________________________________")
```

A. LabelEncoder

   le = LabelEncoder()

   Creates an object of the LabelEncoder class.
B. fit_transform()

df['Purchased'] = le.fit_transform(df['Purchased'])


This function performs two operations:

1. fit():

It identifies the unique categories present in the column.

For example:
['No', 'Yes']

2. transform():

It converts those categories into numerical values.
No  → 0
Yes → 1

fit_transform() combines both operations.

C. inverse_transform()
What if you want to convert numerical values back into their original categories?

Use:
     df['Purchased'] = le.inverse_transform(df['Purchased'])

```python
# Method 2: Ordinal Encoding
# Ordinal Encoding is a categorical encoding technique that converts categorical values into numerical values while maintaining their natural order or ranking.
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Creating dataset
data = {
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha', 'Rohit', 'Anjali', 'Karan', 'Neha'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'City': ['Delhi', 'Mumbai', 'Indore', 'Delhi', 'Mumbai', 'Indore', 'Delhi', 'Mumbai'],
    'Education': ['Graduate', 'Postgraduate', 'Undergraduate', 'Graduate',
                  'Postgraduate', 'Undergraduate', 'Graduate', 'Postgraduate'],
    'Purchased': ['Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print(df)
print("______________________________________________________________________________________________________________________")
# Identify Categorical Columns
cat_cols = df.select_dtypes(include='object').columns

print(cat_cols)
print("______________________________________________________________________________________________________________________")
# Implement Ordinal Encoding
categories = [['Undergraduate', 'Graduate', 'Postgraduate']]
# Creating OrdinalEncoder object
oe = OrdinalEncoder(categories=categories)
# Applying Ordinal Encoding
df[['Education']] = oe.fit_transform(df[['Education']])

print(df)
```

A. OrdinalEncoder()
oe = OrdinalEncoder(categories=categories)


This creates an OrdinalEncoder object.

The categories parameter tells Python the order in which we want to encode our categories.

Method 3: One-Hot Encoding
1. What is One-Hot Encoding?
One-Hot Encoding is a categorical encoding technique that converts categorical values into separate binary (0/1) columns.

```python
# import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Creating dataset
data = {
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha', 'Rohit', 'Anjali', 'Karan', 'Neha'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'City': ['Delhi', 'Mumbai', 'Indore', 'Delhi', 'Mumbai', 'Indore', 'Delhi', 'Mumbai'],
    'Education': ['Graduate', 'Postgraduate', 'Undergraduate', 'Graduate',
                  'Postgraduate', 'Undergraduate', 'Graduate', 'Postgraduate'],
    'Purchased': ['Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Creating OneHotEncoder object
ohe = OneHotEncoder(sparse_output=False)

# Applying One-Hot Encoding
encoded = ohe.fit_transform(df[['City']])

# Converting encoded data into DataFrame
encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(['City'])
)

# Combining encoded columns with original DataFrame
df = pd.concat([df.drop('City', axis=1), encoded_df], axis=1)

print(df)
```
