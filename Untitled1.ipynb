```python
#detect missing values
```

```python
df.isnull().sum()        # Counts missing values in each column

df.isnull().sum().sum()  # Counts total missing values

df.isnull().any()        # Checks whether each column has missing values

df.isna()                # Returns True for missing values
```

```python
#remove missing values
```

```python
df.dropna()              # Removes rows containing at least one missing value

df.dropna(axis=1)         # Removes columns containing missing values

df.dropna(how='all')      # Removes rows where all values are missing

df.dropna(subset=['Age']) # Removes rows where Age is missing
```

```python
#fill missing values
```

```python
#fill with fixed value
df.fillna(0)  # Replaces all missing values with 0
```

```python
#fill with mean
df['Age'].fillna(df['Age'].mean())  # Replaces missing Age with its average
```

```python
#fill with median
df['Age'].fillna(df['Age'].median())  # Replaces missing Age with its median
```

```python
#replace with mode
df['Age'].fillna(df['Age'].mode()[0])  # Replaces missing Age with its most frequent value
```

```python
#replace with forward fill
df.ffill()  # Fills missing values using the previous available value
```

```python
#replace with backward fill
df.bfill()  # Fills missing values using the next available value
```

```python
#scikit learn simplelumper
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')  # Creates a mean imputer

df[['Age']] = imputer.fit_transform(df[['Age']])  # Learns mean and fills missing values
```

```python
#updating the original dataset
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
#removing duplicate records
```

```python
#detect duplicate
df.duplicated()       # Identifies duplicate rows

df.duplicated().sum() # Counts duplicate rows
```

```python
#remove duplicates
df.drop_duplicates()  # Removes duplicate rows
```

```python
#remove duplicate based on a column
df.drop_duplicates(subset=['Name'])  # Removes rows with duplicate names
```

```python
df.drop_duplicates(keep='first')  # Keeps the first occurrence

df.drop_duplicates(keep='last')   # Keeps the last occurrence

df.drop_duplicates(keep=False)    # Removes all occurrences of duplicated rows
```

```python
#to permanently update
df = df.drop_duplicates()
```

```python
#handling incorrect data types
```

```python
df.dtypes  # Checks data types of columns

df['Age'] = df['Age'].astype(int)  # Converts Age into integer

df['Age'] = pd.to_numeric(df['Age'], errors='coerce')  # Converts invalid values into NaN

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')  # Converts values into datetime
```

```python
#remember
astype() – Direct data type conversion.
to_numeric() – Converts values into numerical format.
to_datetime() – Converts values into date-time format.
errors='coerce' – Converts invalid values into missing values instead of throwing an error.
```
