A. Bias

Bias refers to the error caused by making overly simple assumptions about the dataset.

B. Variance

Variance refers to how much a model's predictions change when trained on different datasets.

D. What is Bias-Variance Tradeoff?
It means finding the right balance between bias and variance so that the model performs well on unseen data.

A. Confusion Matrix

A confusion matrix is a table used to evaluate classification models.

It contains four important values:

Term	Meaning

TP (True Positive)	Model predicts Positive, actual is Positive.

TN (True Negative)	Model predicts Negative, actual is Negative.

FP (False Positive)	Model predicts Positive, actual is Negative.

FN (False Negative)	Model predicts Negative, actual is Positive.

B. Important Classification Metrics
1. Accuracy

Measures the percentage of total predictions that are correct.

Accuracy = (TP + TN) / (TP + TN + FP + FN)

2. Precision

Measures how many predicted positive cases were actually positive.

Precision = TP / (TP + FP)

3. Recall

Measures how many actual positive cases were correctly identified.

Recall = TP / (TP + FN)

4. F1-Score

Harmonic mean of Precision and Recall.

F1 = 2 × (Precision × Recall) / (Precision + Recall)

5. Specificity

Measures how many actual negative cases were correctly identified.

Specificity = TN / (TN + FP)




```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix

# Actual values
y_actual = [1, 0, 1, 1, 0, 0, 1, 0]

# Predicted values
y_predicted = [1, 0, 1, 0, 0, 1, 1, 0]

# Calculate evaluation metrics
print("Accuracy:", accuracy_score(y_actual, y_predicted))

print("Precision:", precision_score(y_actual, y_predicted))

print("Recall:", recall_score(y_actual, y_predicted))

print("F1 Score:", f1_score(y_actual, y_predicted))

# Display confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_actual, y_predicted))

```

4. Regression Evaluation Metrics
Regression models predict continuous numerical values.

Examples:
- House price prediction
- Stock price prediction
- Temperature prediction
- Salary prediction
Unlike classification, regression does not predict categories.

It predicts numerical values.
A. Important Regression Metrics
1. MAE (Mean Absolute Error)

Calculates the average absolute difference between actual and predicted values.

MAE = Average of |Actual − Predicted|

Example:

Actual: [100, 200, 300]

Predicted: [110, 190, 280]

Absolute errors: [10, 10, 20]

MAE = 40/3 = 13.33

2. MSE (Mean Squared Error)

Calculates the average of squared differences between actual and predicted values.

MSE = Average of (Actual − Predicted)²

Using the same example:

Squared errors: [100, 100, 400]

MSE = 600/3 = 200

3. RMSE (Root Mean Squared Error)

Square root of MSE.

RMSE = √MSE

RMSE = √200 = 14.14

4. R² Score (R-Squared)

Measures how well the model explains the variation in the target variable.

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

# Actual values
y_actual = [100, 200, 300]

# Predicted values
y_predicted = [110, 190, 280]

# Calculate MAE
print("MAE:", mean_absolute_error(y_actual, y_predicted))

# Calculate MSE
print("MSE:", mean_squared_error(y_actual, y_predicted))

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y_actual, y_predicted))
print("RMSE:", rmse)

# Calculate R2 Score
print("R2 Score:", r2_score(y_actual, y_predicted))
```

5. Cross Validation


Cross Validation is a technique used to evaluate how well a Machine Learning model performs on different portions of a dataset.

Instead of evaluating a model using just one train-test split, we divide the data into multiple parts and evaluate it repeatedly.
B. K-Fold Cross Validation
This is the most commonly used cross-validation technique.

Suppose we have 100 records and choose K = 5.

We divide the dataset into five equal parts.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier

# Load sample dataset
X, y = load_iris(return_X_y=True)

# Create ML model
model = RandomForestClassifier(random_state=42)

# Apply 5-fold cross validation
scores = cross_val_score(model, X, y, cv=5)

# Display individual scores
print("Cross Validation Scores:", scores)

# Calculate average score
print("Average Accuracy:", scores.mean())
```
