Method 1: Feature Creation
Feature Creation means creating a new column (feature) using one or more existing columns in a dataset.

```python
import pandas as pd
import numpy as np

# Creating our dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

# Display original dataset
print(df)
print("---------------------------------------------------------------------------------------------------------------------")

#We will create three new features using existing columns.
# 1: Salary Per Year of Experience
# Creating a new feature by dividing Salary by Experience
df['Salary_Per_Experience'] = df['Salary'] / df['Experience']

print(df)
print("---------------------------------------------------------------------------------------------------------------------")
# 2: Age Category
# Creating age categories using existing Age column
df['Age_Category'] = np.where(df['Age'] < 25, 'Young',
                     np.where(df['Age'] < 30, 'Adult', 'Senior'))

print(df)
print("---------------------------------------------------------------------------------------------------------------------")
# 3: Weekly Working Hours
# Assuming employees work 5 days per week
# Creating a new feature for weekly working hours
df['Weekly_Hours'] = df['Working_Hours'] * 5

print(df)
print("---------------------------------------------------------------------------------------------------------------------")
```

Command	Purpose

df['New_Column']	Creates a new column

np.where()	Creates conditional features

df['A'] + df['B']	Combines two columns

df['A'] / df['B']	Divides two columns

df.head()	Displays first five rows

df.shape	Returns rows and columns


Method 2: Feature Transformation

Feature Transformation is the process of changing the values of existing features (columns) into a different format or scale to make them more suitable for Machine Learning algorithms.

```python
import pandas as pd
import numpy as np

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("---------------------------------------------------------------------------------------------------------------------")
#Different Feature Transformation Techniques
# 1: Mathematical Transformation
# Dividing salary values by 1000
df['Salary'] = df['Salary'] / 1000

# Display updated salary column
print(df['Salary'])
print("---------------------------------------------------------------------------------------------------------------------")

# 2: Square Root Transformation
# Sometimes numerical data contains very large values. Square root transformation reduces the magnitude of those values.
# Applying square root transformation to Salary
df['Salary'] = np.sqrt(df['Salary'])

# Display transformed salary
print(df['Salary'])
print("---------------------------------------------------------------------------------------------------------------------")

# 3: Log Transformation
# # Applying natural logarithm transformation to Salary
df['Salary'] = np.log1p(df['Salary'])

# Display transformed salary
print(df['Salary'])
print("---------------------------------------------------------------------------------------------------------------------")

```

Method 3: Log Transformation

Log Transformation is a technique used to reduce the skewness of numerical data by applying a logarithmic function to its values.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("---------------------------------------------------------------------------------------------------------------------")
# Check the original salary distribution
# Plotting the original salary distribution
plt.hist(df['Salary'], bins=5)

plt.title('Original Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')

plt.show()
print("---------------------------------------------------------------------------------------------------------------------")
# Apply Log Transformation
# Applying log transformation to Salary
df['Salary_Log'] = np.log1p(df['Salary'])

# Display original and transformed values
print(df[['Salary', 'Salary_Log']])
print("---------------------------------------------------------------------------------------------------------------------")
# Visualize the transformed data
# Plotting transformed salary distribution
plt.hist(df['Salary_Log'], bins=5)

plt.title('Log Transformed Salary Distribution')
plt.xlabel('Log Salary')
plt.ylabel('Frequency')

plt.show()
print("---------------------------------------------------------------------------------------------------------------------")
```

Method 4: Binning / Discretization

Binning is a technique in which we convert continuous numerical data into smaller groups or categories.

```python
import pandas as pd
import numpy as np

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("-------------------------------------------------------------------------------------------------------------------")
# Equal Width Binning
# In this technique, we divide numerical values into intervals having equal widths.

# Creating three equal-width age bins
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=3,
    labels=['Young', 'Adult', 'Senior']
)

# Display original age and new category
print(df[['Age', 'Age_Group']])
print("-------------------------------------------------------------------------------------------------------------------")
# Equal Frequency Binning
# In this technique, we divide the data into groups containing approximately the same number of observations.

# Dividing employees into three equal-frequency groups
df['Salary_Group'] = pd.qcut(
    df['Salary'],
    q=3,
    labels=['Low', 'Medium', 'High']
)

# Display original salary and new category
print(df[['Salary', 'Salary_Group']])
print("-------------------------------------------------------------------------------------------------------------------")
```

Command	Purpose

pd.cut()	Performs equal-width binning using a specified number of bins.

pd.qcut()	Performs equal-frequency binning.

bins=3	Divides data into three intervals.

q=3	Divides data into three quantile groups.

labels=[]	Assigns names to the groups.

Method 5: Feature Splitting

Feature Splitting is a technique in which we divide one existing column into two or more separate columns.

```python
import pandas as pd
import numpy as np

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("-------------------------------------------------------------------------------------------------------------------")
# Split Name into First Name and Last Name
# Creating a temporary full name column
df['Full_Name'] = [
    'Aman Sharma',
    'Riya Verma',
    'Rahul Singh',
    'Priya Gupta',
    'Karan Mehta'
]

# Splitting Full_Name into two separate columns
df[['First_Name', 'Last_Name']] = df['Full_Name'].str.split(' ', expand=True)

# Display the split columns
print(df[['Full_Name', 'First_Name', 'Last_Name']])
print("-------------------------------------------------------------------------------------------------------------------")
# Understand the Important Command
df['Full_Name'].str.split(' ', expand=True)
print("-------------------------------------------------------------------------------------------------------------------")
# Extracting regular working hours
df['Regular_Hours'] = df['Working_Hours'].clip(upper=8)

# Extracting overtime hours
df['Overtime_Hours'] = (df['Working_Hours'] - 8).clip(lower=0)

# Display the results
print(df[['Working_Hours', 'Regular_Hours', 'Overtime_Hours']])
#Without expand=True, the result is returned as a Series containing lists.
#With expand=True, the result becomes a DataFrame with separate columns.
```

.str.split()	Splits string values.

.str.get(0)	Extracts the first split element.

.str.get(1)	Extracts the second split element.

.clip()	Restricts values within specified limits.

df['Full_Name']	Selects the Full_Name column.

.str	Allows string operations on column values.

.split(' ')	Splits text wherever a space occurs.

Method 6: Feature Extraction

Feature Extraction is the process of extracting useful information from existing columns and converting it into meaningful features.

```python
import pandas as pd
import numpy as np

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

# Adding joining dates
df['Joining_Date'] = [
    '2022-01-15',
    '2021-06-20',
    '2020-03-10',
    '2023-08-15',
    '2019-11-25'
]

print(df)
print("----------------------------------------------------------------------------------------------------------------------")

# Converting Joining_Date from string to datetime
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])

print(df.dtypes)
print("----------------------------------------------------------------------------------------------------------------------")

# Extracting year from Joining_Date
df['Joining_Year'] = df['Joining_Date'].dt.year

# Extracting month from Joining_Date
df['Joining_Month'] = df['Joining_Date'].dt.month

# Extracting day from Joining_Date
df['Joining_Day'] = df['Joining_Date'].dt.day

print(df[['Joining_Date', 'Joining_Year',
          'Joining_Month', 'Joining_Day']])
print("----------------------------------------------------------------------------------------------------------------------")

# Extracting weekday name
df['Joining_Weekday'] = df['Joining_Date'].dt.day_name()

print(df[['Joining_Date', 'Joining_Weekday']])
print("----------------------------------------------------------------------------------------------------------------------")
# Extract Experience Category from Existing Data
# Creating experience categories
df['Experience_Level'] = np.where(
    df['Experience'] <= 2, 'Fresher',
    np.where(df['Experience'] <= 5, 'Intermediate', 'Experienced')
)

print(df[['Experience', 'Experience_Level']])
print("----------------------------------------------------------------------------------------------------------------------")

```

pd.to_datetime()	Converts data into datetime format.

.dt.year	Extracts year.

.dt.month	Extracts month.

.dt.day	Extracts day.

.dt.day_name()	Extracts weekday name.

.dt.hour	Extracts hour.

.dt.minute	Extracts minute.

.dt.dayofweek	Returns weekday number (Monday = 0).

Method 7: Feature Interaction

Feature Interaction is a technique in which we combine two or more existing features to create a new feature that captures the relationship between them.

```python
import pandas as pd
import numpy as np

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("------------------------------------------------------------------------------------------------------------------")
# Multiplication Interaction
# We will multiply Salary and Experience.
# Creating an interaction feature using multiplication
df['Salary_Experience'] = df['Salary'] * df['Experience']

# Displaying original and new columns
print(df[['Salary', 'Experience', 'Salary_Experience']])
# Addition Interaction
# Adding Age and Experience
df['Age_Experience'] = df['Age'] + df['Experience']

# Displaying the result
print(df[['Age', 'Experience', 'Age_Experience']])
print("------------------------------------------------------------------------------------------------------------------")
# Division Interaction
# Calculating salary per working hour
df['Salary_Per_Hour'] = df['Salary'] / (df['Working_Hours'] * 5)

# Displaying the result
print(df[['Salary', 'Working_Hours', 'Salary_Per_Hour']])
print("------------------------------------------------------------------------------------------------------------------")
```

Method 8: Feature Scaling

Feature Scaling is a technique used to bring numerical features into a similar range or scale.
. Types of Feature Scaling
We will study three important techniques:
1. Min-Max Scaling (Normalization)
2. Standardization (Z-score Scaling)
3. Robust Scaling

```python
import pandas as pd
import numpy as np

# Importing scaling techniques
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
# Selecting numerical columns
numerical_data = df[['Age', 'Salary', 'Experience', 'Working_Hours']]
print("-----------------------------------------------------------------------------------------------------------------")
# Min-Max Scaling (Normalization)
# Min-Max Scaling converts numerical values into a specified range, generally between 0 and 1.
# Creating MinMaxScaler object
scaler = MinMaxScaler()

# Applying Min-Max Scaling
scaled_data = scaler.fit_transform(numerical_data)

# Converting the result into a DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=numerical_data.columns
)

# Displaying scaled values
print(scaled_df)
print("-----------------------------------------------------------------------------------------------------------------")
# Standardization (Z-score Scaling)

# Standardization transforms numerical data so that:
# - Mean becomes approximately 0.
# - Standard deviation becomes approximately 1.
# Creating StandardScaler object
scaler = StandardScaler()

# Applying standardization
scaled_data = scaler.fit_transform(numerical_data)

# Converting result into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=numerical_data.columns
)

# Displaying standardized values
print(scaled_df)
print("-----------------------------------------------------------------------------------------------------------------")
# Robust Scaling

# Robust Scaling is a technique that scales numerical data using:
# - Median
# - Interquartile Range (IQR)
# Creating RobustScaler object
scaler = RobustScaler()

# Applying Robust Scaling
scaled_data = scaler.fit_transform(numerical_data)

# Converting result into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=numerical_data.columns
)

# Displaying scaled values
print(scaled_df)
print("-----------------------------------------------------------------------------------------------------------------")
```

```python
# Never fit your scaler on the complete dataset before splitting it into training and testing data.
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Selecting features
X = df[['Age', 'Salary', 'Experience', 'Working_Hours']]

# Splitting dataset into training and testing sets
X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=42
)

# Creating scaler
scaler = StandardScaler()

# Learning parameters from training data only
X_train_scaled = scaler.fit_transform(X_train)

# Applying the same parameters to testing data
X_test_scaled = scaler.transform(X_test)
```

MinMaxScaler()	Creates a Min-Max scaler object.

fit_transform()	Learns minimum and maximum values and transforms the data.

fit()	Learns scaling parameters.

transform()	Applies the learned transformation.

StandardScaler()	Creates a standardization object.

mean_	Displays learned column means.

scale_	Displays learned scaling factors (standard deviations).

Method 9: Feature Selection

Feature Selection is the process of selecting the most important and relevant columns from a dataset while removing unnecessary or irrelevant columns.

```python
# Method A: Filter Method
# Filter methods select features using statistical measurements without training a Machine Learning model.

# We will study three techniques:
# 1. Variance Threshold
# 2. Correlation
# 3. Statistical Feature Selection
import pandas as pd
import numpy as np

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("-------------------------------------------------------------------------------------------------------")
# Technique 1: Variance Threshold
# Variance Threshold removes columns that have very little variation in their values.
from sklearn.feature_selection import VarianceThreshold

# Selecting numerical columns
X = df[['Age', 'Salary', 'Experience', 'Working_Hours']]

# Creating Variance Threshold object
selector = VarianceThreshold(threshold=0)

# Applying feature selection
selected_data = selector.fit_transform(X)

# Getting selected column names
selected_columns = X.columns[selector.get_support()]

print(selected_columns)
print("------------------------------------------------------------------------------------------------------------------")
# Technique 2: Correlation

# Correlation measures the relationship between two numerical columns.
#Its value ranges from -1 to +1.
# Calculating correlation between numerical columns
correlation = df[['Age', 'Salary', 'Experience',
                  'Working_Hours']].corr()

# Displaying correlation matrix
print(correlation)
print("------------------------------------------------------------------------------------------------------------------")
import seaborn as sns
import matplotlib.pyplot as plt

# Creating correlation heatmap
sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title('Feature Correlation')
plt.show()
print("------------------------------------------------------------------------------------------------------------------")
# Technique 3: Statistical Feature Selection
# This technique uses statistical tests to identify features that are useful for predicting a target variable.
# Adding a target column to our existing dataset
df['Performance'] = [0, 1, 1, 1, 1]

# 0 = Low Performance
# 1 = High Performance

print(df)
print("------------------------------------------------------------------------------------------------------------------")
from sklearn.feature_selection import SelectKBest, f_classif

# Selecting numerical input features
X = df[['Age', 'Salary', 'Experience', 'Working_Hours']]

# Selecting target variable
y = df['Performance']

# Selecting the best 2 features
selector = SelectKBest(score_func=f_classif, k=2)

# Applying feature selection
selected_data = selector.fit_transform(X, y)

# Displaying selected feature names
selected_columns = X.columns[selector.get_support()]

print(selected_columns)
print("------------------------------------------------------------------------------------------------------------------")
```

```python
# Wrapper Method

# Wrapper methods select features by repeatedly training a Machine Learning model with different combinations of features.
# Recursive Feature Elimination (RFE)
# RFE repeatedly removes the least important features until the desired number of features remains.
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier

# Selecting input features
X = df[['Age', 'Salary', 'Experience', 'Working_Hours']]

# Selecting target
y = df['Performance']

# Creating Random Forest model
model = RandomForestClassifier(random_state=42)

# Selecting the best 2 features
selector = RFE(model, n_features_to_select=2)

# Applying RFE
selector.fit(X, y)

# Displaying selected features
print(X.columns[selector.support_])
print("------------------------------------------------------------------------------------------------------------------")
```

```python
# Embedded Method
# Embedded methods perform feature selection during the training process of a Machine Learning model.
from sklearn.ensemble import RandomForestClassifier

# Selecting input features
X = df[['Age', 'Salary', 'Experience', 'Working_Hours']]

# Selecting target
y = df['Performance']

# Creating Random Forest model
model = RandomForestClassifier(random_state=42)

# Training the model
model.fit(X, y)

# Getting feature importance scores
importance = model.feature_importances_

# Creating a DataFrame for feature importance
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': importance
})

# Sorting features by importance
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print(feature_importance)
```


VarianceThreshold()	Creates feature selection object.

get_support()	Returns True/False for selected features.

RFE()	Creates Recursive Feature Elimination object.

n_features_to_select=2	Keeps two features.

selector.fit()	Performs feature selection.

selector.support_	Identifies selected features.

selector.ranking_	Displays feature rankings.

RandomForestClassifier()	Creates Random Forest model.

model.fit()	Trains the model.

sort_values()	Sorts features by importance.

Method 10: Dimensionality Reduction
We will use the same employee dataset that we have been using throughout Feature Engineering.

Dimensionality Reduction is the process of reducing the number of features (columns) in a dataset while preserving as much important information as possible.

```python
import pandas as pd
import numpy as np

# Importing PCA
from sklearn.decomposition import PCA

# Importing StandardScaler
from sklearn.preprocessing import StandardScaler

# Creating the same employee dataset
data = {
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Age': [22, 25, 30, 28, 35],
    'Salary': [25000, 40000, 50000, 45000, 60000],
    'Experience': [1, 3, 5, 4, 8],
    'Working_Hours': [8, 9, 10, 8, 9]
}

df = pd.DataFrame(data)

print(df)
print("-----------------------------------------------------------------------------------------------------------")
# Selecting numerical columns
X = df[['Age', 'Salary', 'Experience', 'Working_Hours']]

print(X)
# Apply Feature Scaling
# Creating StandardScaler object
scaler = StandardScaler()

# Scaling numerical features
X_scaled = scaler.fit_transform(X)

print(X_scaled)
print("-----------------------------------------------------------------------------------------------------------")
# Apply PCA
# Creating PCA object with 2 components
pca = PCA(n_components=2)

# Applying PCA to scaled data
X_pca = pca.fit_transform(X_scaled)

# Converting PCA output into DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=['PC1', 'PC2']
)

# Displaying reduced dataset
print(pca_df)
print("-----------------------------------------------------------------------------------------------------------")
# Check Explained Variance
# Checking variance captured by each component
print(pca.explained_variance_ratio_)
# Check Total Explained Variance
# Calculating total variance captured
total_variance = sum(pca.explained_variance_ratio_)

print("Total Explained Variance:", total_variance)
```


PCA()	Creates a PCA object.

n_components=2	Selects two principal components.

fit_transform()	Learns PCA directions and transforms data.

explained_variance_ratio_	Returns variance captured by each component.

components_	Returns principal component directions.

inverse_transform()	Reconstructs approximate original data from components.
