```python
#feature engineering
#Feature Engineering is the process of creating new features, modifying existing features, or selecting useful features from an existing
#dataset to improve Machine Learning model performance.
import pandas as pd
import numpy as np

data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Age': [18, 19, 20, 21, 22, 19, 20, 21, 23, 18],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Marks': [45, 78, 90, 55, 88, 65, 72, 95, 40, 82],
    'Study_Hours': [2, 4, 6, 3, 5, 3, 4, 7, 1, 5],
    'Salary': [15000, 22000, 30000, 18000, 28000,
               20000, 25000, 35000, 12000, 27000]
}

df = pd.DataFrame(data)

df
```

```python
import pandas as pd
import numpy as np
#create dataset copy
df_fe = df.copy()
print("copy of df=",df_fe)
print("_______________________________________________________________________________________________________________________________")

#Feature Creation
#Create new columns using existing columns
df_fe['Total_Score'] = df_fe['Marks'] + df_fe['Study_Hours']
df_fe['Marks_Per_Hour'] = df_fe['Marks'] / df_fe['Study_Hours']
print("df_fe['Total_Score']=",df_fe['Total_Score'])
print("_______________________________________________________________________________________________________________________________")
print("df_fe['marks_per_hour]=",df_fe['Marks_Per_Hour'])
print("_______________________________________________________________________________________________________________________________")

#Feature Binning
#Convert numerical values into categories
df_fe['Age_Group'] = pd.cut(
    df_fe['Age'],
    bins=[0, 19, 21, 100],
    labels=['Teenager', 'Young Adult', 'Adult']
)
print("numeric to categories=",df_fe['Age_Group'])
print("_______________________________________________________________________________________________________________________________")

df_fe['Performance'] = pd.cut(
    df_fe['Marks'],
    bins=[0, 60, 80, 100],
    labels=['Low', 'Average', 'High']
)
print("numeric to categories=",df_fe['Performance'])
print("_______________________________________________________________________________________________________________________________")

#Feature Transformation
#Log Transformation is a Feature Engineering technique in which we apply a logarithm to numerical values
#to reduce skewness and make highly uneven data more balanced.
df_fe['Salary_Log'] = np.log1p(df_fe['Salary']) #log transformation
print("log transformation=",df_fe['Salary_Log'])
print("_______________________________________________________________________________________________________________________________")

df_fe['Age_Squared'] = df_fe['Age'] ** 2 #square transformation
print("squared transformation=",df_fe['Age_Squared'])
print("_______________________________________________________________________________________________________________________________")

#Feature Interaction
#Combine two existing columns
df_fe['Age_Marks'] = df_fe['Age'] * df_fe['Marks']
print("combine column=",df_fe['Age_Marks'])
print("_______________________________________________________________________________________________________________________________")

#Feature Scaling
#Bringing numerical values to a comparable scale.
#Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_fe[['Age_Standard', 'Salary_Standard']] = scaler.fit_transform(
    df_fe[['Age', 'Salary']]
)
print("feature scaling,standardization=",df_fe[['Age_Standard', 'Salary_Standard']])
print("_______________________________________________________________________________________________________________________________")

#normalization
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_fe[['Age_Normalized', 'Salary_Normalized']] = scaler.fit_transform(
    df_fe[['Age', 'Salary']]
)
print("normalization=",df_fe[['Age_Normalized', 'Salary_Normalized']])
print("_______________________________________________________________________________________________________________________________")

#Feature Extraction
#Extracting useful information from existing data.
#For example, create a new category based on marks.
#Extract Performance Category
df_fe['Grade'] = np.where(
    df_fe['Marks'] >= 80, 'A',
    np.where(df_fe['Marks'] >= 60, 'B', 'C')
)
print("extraction =",df_fe['Grade'])
print("_______________________________________________________________________________________________________________________________")

#Feature Selection
#Selecting only useful columns for Machine Learning.
#Select Specific Features
selected_features = df_fe[
    ['Age', 'Marks', 'Study_Hours', 'Salary']
]
print("selection =",selected_features)
print("_______________________________________________________________________________________________________________________________")

#display selected feature
selected_features.head()
print("_______________________________________________________________________________________________________________________________")

#display final result
df_fe.head(10)
```
