```python
#feature engineering
#Feature Engineering is the process of creating new features, modifying existing features, or selecting useful features from an existing dataset to improve
# Machine Learning model performance.
import pandas as pd
import numpy as np

data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Age': [18, 19, 20, 21, 22, 19, 20, 21, 23, 18],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Marks': [45, 78, 90, 55, 88, 65, 72, 95, 40, 82],
    'Study_Hours': [2, 4, 6, 3, 5, 3, 4, 7, 1, 5],
    'Salary': [15000, 22000, 30000, 18000, 28000,
               20000, 25000, 35000, 12000, 27000]
}

df = pd.DataFrame(data)

df
```

```python
import pandas as pd
import numpy as np
#create dataset copy
df_fe = df.copy()

#Feature Creation
#Create new columns using existing columns
df_fe['Total_Score'] = df_fe['Marks'] + df_fe['Study_Hours']
df_fe['Marks_Per_Hour'] = df_fe['Marks'] / df_fe['Study_Hours']

#Feature Binning
#Convert numerical values into categories
df_fe['Age_Group'] = pd.cut(
    df_fe['Age'],
    bins=[0, 19, 21, 100],
    labels=['Teenager', 'Young Adult', 'Adult']
)

df_fe['Performance'] = pd.cut(
    df_fe['Marks'],
    bins=[0, 60, 80, 100],
    labels=['Low', 'Average', 'High']
)

#Feature Transformation
df_fe['Salary_Log'] = np.log1p(df_fe['Salary']) #log transformation

df_fe['Age_Squared'] = df_fe['Age'] ** 2 #square transformation

#Feature Interaction
#Combine two existing columns
df_fe['Age_Marks'] = df_fe['Age'] * df_fe['Marks']

#Feature Encoding
#Convert categorical data into numerical values
#label encoding
df_fe['Performance_Encoded'] = df_fe['Performance'].map({
    'Low': 0,
    'Average': 1,
    'High': 2
})

#one hot encoding
df_fe = pd.get_dummies(df_fe, columns=['Age_Group'], dtype=int)

#Feature Scaling
#Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_fe[['Age_Scaled', 'Salary_Scaled']] = scaler.fit_transform(
    df_fe[['Age', 'Salary']]
)

#normalization
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_fe[['Age_Normalized', 'Salary_Normalized']] = scaler.fit_transform(
    df_fe[['Age', 'Salary']]
)
#Feature Selection
#Select specific useful columns
selected_features = df_fe[['Age', 'Salary', 'Marks', 'Study_Hours']]
selected_features.head()

#display final result
df_fe.head(10)
```
